```python
import gradio as gr
```

```python
def chatbot_response(message, history):

    message = message.lower()

    if "hello" in message or "hi" in message:
        return "Hello! Welcome to my AI chatbot. How can I help you?"

    elif "artificial intelligence" in message or "what is ai" in message:
        return "Artificial Intelligence (AI) is a technology that enables computers to perform tasks that normally require human intelligence."

    elif "machine learning" in message:
        return "Machine Learning is a branch of AI that enables computers to learn patterns from data and make predictions."

    elif "deep learning" in message:
        return "Deep Learning is a part of Machine Learning that uses multi-layered neural networks."

    elif "gradio" in message:
        return "Gradio is a Python library used to build interactive web applications for machine learning models."

    else:
        return "Thank you for your question! I am a demo chatbot. Please try asking about AI, Machine Learning, Deep Learning, or Gradio."
```

```python
demo = gr.ChatInterface(
    fn=chatbot_response,

    title="Interactive AI Chatbot",

    description=(
        "Welcome to my AI Chatbot! "
        "Ask questions about Artificial Intelligence, "
        "Machine Learning, Deep Learning, and Gradio. "
        "Type your question below or select a sample prompt."
    ),

    examples=[
        "What is Artificial Intelligence?",
        "Explain Machine Learning",
        "What is Deep Learning?",
        "What is Gradio?"
    ]
)
```

```python
demo.launch()
```
