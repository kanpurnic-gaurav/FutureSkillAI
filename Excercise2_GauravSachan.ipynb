```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img1 = cv2.imread("image1.jpg")
img2 = cv2.imread("image2.jpg")

print("Image 1 loaded:", img1 is not None)
print("Image 2 loaded:", img2 is not None)
```

```python
print("Image 1 dimensions:", img1.shape)
print("Image 2 dimensions:", img2.shape)
```

```python
img1 = cv2.resize(img1, (500, 400))
img2 = cv2.resize(img2, (500, 400))

print("New Image 1 dimensions:", img1.shape)
print("New Image 2 dimensions:", img2.shape)
```

```python
blended = cv2.addWeighted(
    img1, 0.5,
    img2, 0.5,
    0
)
```

```python
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Original Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite("original_image1.jpg", img1)
cv2.imwrite("original_image2.jpg", img2)
cv2.imwrite("blended_image.jpg", blended)

```
