```python

import re
import pandas as pd
```

```python

def mask_pii(text):

    # Mask labelled home addresses
    text = re.sub(
        r'(?im)\b(?:home\s+)?address\s*[:\-]\s*.*$',
        '[ADDRESS_MASKED]',
        text
    )

    # Mask email addresses
    text = re.sub(
        r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
        '[EMAIL_MASKED]',
        text
    )

    # Mask Indian mobile phone numbers
    text = re.sub(
        r'\b(?:\+91[\s-]?)?[6-9]\d{4}[\s-]?\d{5}\b',
        '[PHONE_MASKED]',
        text
    )

    # Mask 12-digit Aadhaar-like numbers
    text = re.sub(
        r'\b\d{4}[\s-]?\d{4}[\s-]?\d{4}\b',
        '[AADHAAR_MASKED]',
        text
    )

    # Mask PAN-like numbers
    text = re.sub(
        r'\b[A-Z]{5}\d{4}[A-Z]\b',
        '[PAN_MASKED]',
        text
    )

    return text
```

```python

sample_text = """
My name is Example Citizen.
Please contact me at 9876543210.
My email is example.citizen@example.com.
Aadhaar: 1234 5678 9012
PAN: ABCDE1234F
Address: 24 Green Park Road, Kanpur, Uttar Pradesh
"""

print("Original Complaint:")
print(sample_text)

print("\nMasked Complaint:")
print(mask_pii(sample_text))
```

```python
import pandas as pd

# Load previously saved DataFrame
df = pd.read_csv("Jan_Samadhan_Cleaned_Data.csv")

# Apply PII masking
df["Masked_Complaint"] = (
    df["Complaint_Description"]
    .fillna("")
    .apply(mask_pii)
)

# Display results
display(df[["Complaint_Description", "Masked_Complaint"]].head())
```

```python

# Create a separate masked column
df["Masked_Complaint"] = (
    df["Complaint_Description"].apply(mask_pii)
)

# Display original and masked complaints
print(df[[
    "Complaint_Description",
    "Masked_Complaint"
]].head(10))
```

```python

df.to_csv(
    "Jan_Samadhan_Masked_Data.csv",
    index=False
)

print("Masked dataset saved successfully!")
```
