```python
from sklearn.linear_model import LinearRegression
```

```python
# 1. Input Data (Must be 2D array for sklearn, Target is 1D)
X_pages = [[10], [20], [30], [40]]  # Page counts
y_days = [2, 4, 6, 8]             # Actual processing days

# 2. Train the Model
model = LinearRegression()
model.fit(X_pages, y_days)

# 3. Predict for a new file with 50 pages
new_file = [[50]]
predicted_days = model.predict(new_file)

print(f"Predicted days for 50 pages: {predicted_days[0]:.1f} days")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

X_pages = np.array([[10], [20], [30], [40]])
y_days = np.array([2, 4, 6, 8])

model = LinearRegression().fit(X_pages, y_days)

# Smooth line across 10 to 50 pages
x_line = np.linspace(10, 50, 100).reshape(-1, 1)
y_line = model.predict(x_line)

plt.figure(figsize=(7, 4))
plt.scatter(X_pages, y_days, color="tab:blue", label="Observed data")
plt.plot(x_line, y_line, color="tab:orange", label="Fitted line (y = 0.2x)")
plt.scatter([[50]], model.predict([[50]]), color="tab:green", zorder=5, label="50-page prediction (10d)")
plt.xlabel("Page Count")
plt.ylabel("Processing Days")
plt.title("Linear Regression: Pages vs. Processing Days")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
