```python

import pandas as pd

risk_data = {
    "Risk": [
        "Prompt Injection",
        "Algorithmic Bias"
    ],

    "Description": [
        "Malicious instructions embedded in citizen complaints may manipulate AI responses.",
        "AI may produce unfair predictions for certain languages, regions or citizen groups."
    ],

    "Potential_Impact": [
        "Unauthorized information disclosure or incorrect responses.",
        "Incorrect grievance classification and unequal service delivery."
    ],

    "Risk_Level": [
        "High",
        "High"
    ],

    "NIC_Mitigation": [
        "Treat complaint text as untrusted input, use secure prompts, restrict access and validate AI outputs.",
        "Use diverse training data, test performance across languages and regions, conduct regular audits and provide human review."
    ]
}

risk_df = pd.DataFrame(risk_data)

print(risk_df)
```

```python

import matplotlib.pyplot as plt

risk_counts = risk_df["Risk_Level"].value_counts()

plt.figure(figsize=(7, 5))

risk_counts.plot(
    kind="bar",
    title="Jan-Samadhan AI Risk Assessment"
)

plt.xlabel("Risk Level")
plt.ylabel("Number of Risks")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("Jan_Samadhan_Risk_Assessment.png")
plt.show()
```

```python

risk_df.to_csv(
    "Jan_Samadhan_Risk_Assessment.csv",
    index=False
)

print("Risk assessment saved successfully!")
```
